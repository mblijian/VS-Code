```python
def pick_up():
    import time
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(0,0,-75,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)
    got.mechanical_joint_control(90,30,30,1000)

def put_down():
    got.mechanical_joint_control(0,0,-75,1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)
```

```python
import random
player_hand = []
dealer_hand = []

# give two cards to player and dealer
for i in range(2):
    player_hand.append(random.randint(1,10))
    dealer_hand.append(random.randint(1,10))
# sort the cards
player_hand.sort()
dealer_hand.sort()

print(f"Your cards: {player_hand}")
print(f"Dealer cards: {dealer_hand}")

while sum(player_hand)<21:
    choice = input("Hit or stand? Enter h or s.").lower()
    if choice == "h":
        new_card = random.randint(1,10)
        player_hand.append(new_card)
        player_hand.sort()

        print(f"Your new card is : {new_card}")
        print(f"Your cards:{player_hand}")
        print(f"Your total points:{sum(player_hand)}")
    elif choice == "s":
        print("You chose to stand.")
        break
    else:
        print("Please enter h or s.")

# compare the player hand with the dealer hand.
if sum(player_hand)> 21:
    print("You went over 21, you lose!")
else:
    print(f"Dealer's cards:{dealer_hand}")

    while sum(dealer_hand)<17:
        new_card = random.randint(1,10)
        dealer_hand.append(new_card)
        dealer_hand.sort()
        
        print(f"Dealer new card is : {new_card}")
        print(f"Dealer cards:{dealer_hand}")
        print(f"Dealer total points:{sum(dealer_hand)}")
    #show final result
    player_total = sum(player_hand)
    dealer_total = sum(dealer_hand)

    print(f"Your cards:{player_hand}")
    print(f"Your total points:{player_total}")
    print(f"Dealer cards:{dealer_hand}")
    print(f"Dealer total points:{dealer_total}")

    # Decide the winner:
    if dealer_total>21:
        print("The dealer hand went over 21, you win!")
    elif player_total> dealer_total:
        print("You win!")
    elif player_total< dealer_total:
        print("You lose!")
    else:
        print("It is a draw!")
```

```python
import random

player_health = 30
monster_health = 30

print("Welcome to Monster Battle!")

while player_health > 0 and monster_health > 0:

    print(f"\nYour health: {player_health}")
    print(f"Monster health: {monster_health}")

    print("\nChoose an action:")
    print("1. Attack")
    print("2. Heal")
    print("3. Defend")

    choice = input("Enter 1, 2, or 3: ")

    defending = False

    if choice == "1":
        player_damage = random.randint(5, 10)
        monster_health -= player_damage

        print(f"You attacked the monster for {player_damage} damage!")

    elif choice == "2":
        healing = random.randint(4, 8)
        player_health += healing

        print(f"You recovered {healing} health points!")

    elif choice == "3":
        defending = True
        print("You raised your shield!")

    else:
        print("Please enter 1, 2, or 3.")
        continue

    # The monster attacks only if it is still alive
    if monster_health > 0:
        monster_damage = random.randint(3, 8)

        if defending:
            monster_damage = monster_damage // 2
            print("Your shield reduced the monster's damage!")

        player_health -= monster_damage

        print(f"The monster attacked you for {monster_damage} damage!")

# Show the result
if player_health <= 0:
    print("\nThe monster defeated you!")
else:
    print("\nYou defeated the monster!")
```
