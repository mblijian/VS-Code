3. face finder

The face finder gives back a list of faces. Each face is `(x,y,w,h)`:
`x,y` is the top left corner, `w` is the width, `h` is the height.

```python
finder = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")

def find_faces(frame):
    gray = cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    faces = finder.detectMultiScale(gray,1.3,5)
    return faces
```

```python
cap = cv2.VideoCapture(0)
time.sleep(1)
ok,frame = cap.read()
cap.release()

faces = find_faces(frame)
print(f"I can see {len(faces)} face(s).")
for x,y,w,h in faces:
    print(f"x: {x}, y: {y}, width: {w}, height: {h}")
```

4. Filter Booth with color keys

```python
my_name = "Alice"
COLORS = {"r":(0,0,255),"g":(0,255,0),"b":(255,0,0),"y":(0,255,255)}
color = COLORS["g"]

cap = cv2.VideoCapture(0)
print("Press r, g, b, y to change the color. Press q to stop.")

while True:
    ok,frame = cap.read()
    if not ok:
        break
    frame = cv2.flip(frame,1)

    for x,y,w,h in find_faces(frame):
        cv2.rectangle(frame,(x,y),(x+w,y+h),color,3)
        cv2.putText(frame,my_name,(x,y-10),cv2.FONT_HERSHEY_SIMPLEX,1,color,3)

    cv2.imshow("Filter Booth",frame)
    key = chr(cv2.waitKey(1) & 0xFF)
    if key == "q":
        break
    if key in COLORS:
        color = COLORS[key]

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
```

5. stickers

Each sticker is a function. It draws on the face using `x,y,w,h`.

```python
def draw_hat(frame,x,y,w,h,color):
    cv2.rectangle(frame,(x+w//6,y-h//2),(x+w-w//6,y),color,-1)      # top of the hat
    cv2.rectangle(frame,(x-w//6,y-h//10),(x+w+w//6,y),color,-1)     # brim of the hat

def draw_glasses(frame,x,y,w,h,color):
    left = (x+w//3,y+h*2//5)
    right = (x+w*2//3,y+h*2//5)
    cv2.circle(frame,left,w//7,color,4)
    cv2.circle(frame,right,w//7,color,4)
    cv2.line(frame,(x+w//3+w//7,y+h*2//5),(x+w*2//3-w//7,y+h*2//5),color,4)

def draw_crown(frame,x,y,w,h,color):
    cv2.rectangle(frame,(x,y-h//5),(x+w,y),color,-1)
    cv2.circle(frame,(x+w//6,y-h//4),w//10,color,-1)
    cv2.circle(frame,(x+w//2,y-h//3),w//10,color,-1)
    cv2.circle(frame,(x+w-w//6,y-h//4),w//10,color,-1)

def draw_nose(frame,x,y,w,h,color):
    cv2.circle(frame,(x+w//2,y+h*3//5),w//8,color,-1)
```

```python
my_name = "Alice"
COLORS = {"r":(0,0,255),"g":(0,255,0),"b":(255,0,0),"y":(0,255,255)}
STICKERS = {"1":"hat","2":"glasses","3":"crown","4":"nose","0":"none"}
color = COLORS["y"]
sticker = "none"

cap = cv2.VideoCapture(0)
print("Colors: r, g, b, y. Stickers: 1, 2, 3, 4, 0. Press q to stop.")

while True:
    ok,frame = cap.read()
    if not ok:
        break
    frame = cv2.flip(frame,1)

    # sticker block
    for x,y,w,h in find_faces(frame):
        if sticker == "hat":
            draw_hat(frame,x,y,w,h,color)
        elif sticker == "glasses":
            draw_glasses(frame,x,y,w,h,color)
        elif sticker == "crown":
            draw_crown(frame,x,y,w,h,color)
        elif sticker == "nose":
            draw_nose(frame,x,y,w,h,color)
        cv2.putText(frame,my_name,(x,y+h+30),cv2.FONT_HERSHEY_SIMPLEX,1,color,3)

    cv2.putText(frame,f"Sticker: {sticker}",(10,30),cv2.FONT_HERSHEY_SIMPLEX,0.8,(255,255,255),2)
    cv2.imshow("Filter Booth",frame)

    # keyboard block
    key = chr(cv2.waitKey(1) & 0xFF)
    if key == "q":
        break
    if key in COLORS:
        color = COLORS[key]
    if key in STICKERS:
        sticker = STICKERS[key]

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
```

## Session 5

1. did I catch it?

The star is caught when its point is **inside** the face box.

```python
def is_inside(point,box):
    px,py = point
    x,y,w,h = box
    return x < px < x+w and y < py < y+h

print(is_inside((150,150),(100,100,200,200)))     # True
print(is_inside((50,400),(100,100,200,200)))      # False
```

2. catch the star and count the score

```python
star = {"x":300,"y":0,"speed":6,"color":(0,255,255)}
score = 0

cap = cv2.VideoCapture(0)
print("Catch the star with your face! Press q to stop.")

while True:
    ok,frame = cap.read()
    if not ok:
        break
    frame = cv2.flip(frame,1)
    height,width,channels = frame.shape

    # star block
    star["y"] = star["y"] + star["speed"]
    if star["y"] > height:
        star["y"] = 0
        star["x"] = random.randint(20,width-20)
    cv2.circle(frame,(star["x"],star["y"]),15,star["color"],-1)

    # catch block
    for face in find_faces(frame):
        x,y,w,h = face
        cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),3)
        if is_inside((star["x"],star["y"]),face):
            score = score + 1
            star["y"] = 0
            star["x"] = random.randint(20,width-20)

    cv2.putText(frame,f"Score: {score}",(10,40),cv2.FONT_HERSHEY_SIMPLEX,1.2,(255,255,255),3)
    cv2.imshow("Star Catcher",frame)
    key = chr(cv2.waitKey(1) & 0xFF)
    if key == "q":
        break

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
print(f"Your score is {score}.")
```

2. star types

Each type has a color, a speed and points. The bomb takes points away!

```python
STAR_TYPES = [
    {"name":"gold","color":(0,215,255),"speed":6,"points":1},
    {"name":"diamond","color":(255,255,0),"speed":11,"points":3},
    {"name":"bomb","color":(0,0,255),"speed":7,"points":-2}
]

def make_star(width):
    kind = random.choice(STAR_TYPES)
    star = {"x":random.randint(20,width-20),"y":0,
            "name":kind["name"],"color":kind["color"],
            "speed":kind["speed"],"points":kind["points"]}
    return star

print(make_star(640))
print(make_star(640))
```

3. many stars

```python
NUMBER_OF_STARS = 4
stars = []
score = 0

cap = cv2.VideoCapture(0)
print("Catch gold and diamonds, stay away from the red bombs! Press q to stop.")

while True:
    ok,frame = cap.read()
    if not ok:
        break
    frame = cv2.flip(frame,1)
    height,width,channels = frame.shape

    # make new stars block
    while len(stars) < NUMBER_OF_STARS:
        stars.append(make_star(width))

    faces = find_faces(frame)
    for x,y,w,h in faces:
        cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),3)

    # star block
    for star in stars.copy():             # loop over a copy, because we remove stars
        star["y"] = star["y"] + star["speed"]
        cv2.circle(frame,(star["x"],star["y"]),15,star["color"],-1)

        caught = False
        for face in faces:
            if is_inside((star["x"],star["y"]),face):
                caught = True

        if caught:
            score = score + star["points"]
            stars.remove(star)
        elif star["y"] > height:
            stars.remove(star)

    cv2.putText(frame,f"Score: {score}",(10,40),cv2.FONT_HERSHEY_SIMPLEX,1.2,(255,255,255),3)
    cv2.imshow("Star Catcher",frame)
    key = chr(cv2.waitKey(1) & 0xFF)
    if key == "q":
        break

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
print(f"Your score is {score}.")
```

## Session 8

1. BGR and HSV

The camera gives us BGR. To **find** a color, HSV is easier:
- **H** (hue): which color? 0 - 179
- **S** (saturation): how strong is the color? 0 - 255
- **V** (value): how bright? 0 - 255

```python
blue_bgr = np.uint8([[[255,0,0]]])                # one blue pixel
blue_hsv = cv2.cvtColor(blue_bgr,cv2.COLOR_BGR2HSV)
h,s,v = map(int,blue_hsv[0][0])
print(f"Blue in BGR is (255,0,0). In HSV it is ({h},{s},{v}).")
```

Put your wand in the small circle in the middle of the screen. Write down the H, S and V numbers.

5. find the wand

This function gives back the center of the wand as a tuple `(x,y)`. If there is no wand, it gives back `None`.

```python
def find_wand(frame,wand):
    lower,upper = WANDS[wand]
    hsv = cv2.cvtColor(frame,cv2.COLOR_BGR2HSV)
    mask = cv2.inRange(hsv,lower,upper)
    shapes,levels = cv2.findContours(mask,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)

    if len(shapes) == 0:
        return None
    biggest = max(shapes,key=cv2.contourArea)     # the biggest white shape
    if cv2.contourArea(biggest) < 500:            # too small, it is only noise
        return None
    center,radius = cv2.minEnclosingCircle(biggest)
    x,y = map(int,center)
    return (x,y)
```

```python
wand = "blue"

cap = cv2.VideoCapture(0)
print("Show your wand to the camera. Press q to stop.")

while True:
    ok,frame = cap.read()
    if not ok:
        break
    frame = cv2.flip(frame,1)

    point = find_wand(frame,wand)
    if point is None:
        cv2.putText(frame,"No wand",(10,40),cv2.FONT_HERSHEY_SIMPLEX,1.2,(0,0,255),3)
    else:
        cv2.circle(frame,point,20,(255,255,255),3)
        cv2.putText(frame,f"Wand at {point}",(10,40),cv2.FONT_HERSHEY_SIMPLEX,1.2,(0,255,0),3)

    cv2.imshow("Magic Wand",frame)
    key = chr(cv2.waitKey(1) & 0xFF)
    if key == "q":
        break

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
```

3. the full Magic Wand Painter

Now every dot remembers its own color and size, so every dot is a **dictionary**.

```python
wand = "blue"
PENS = {"r":(0,0,255),"g":(0,255,0),"y":(0,255,255),"w":(255,255,255),"p":(255,0,255)}
SIZES = {"1":4,"2":8,"3":16}
pen = PENS["y"]
size = SIZES["2"]
trail = []

cap = cv2.VideoCapture(0)
print("Pens: r, g, y, w, p. Sizes: 1, 2, 3. Press c to clear, q to stop.")

while True:
    ok,frame = cap.read()
    if not ok:
        break
    frame = cv2.flip(frame,1)

    # wand block
    point = find_wand(frame,wand)
    dot = {"point":point,"color":pen,"size":size}
    trail.append(dot)

    # drawing block
    for i in range(1,len(trail)):
        old = trail[i-1]
        new = trail[i]
        if old["point"] is not None and new["point"] is not None:
            cv2.line(frame,old["point"],new["point"],new["color"],new["size"])

    if point is not None:
        cv2.circle(frame,point,12,pen,-1)
    cv2.putText(frame,f"Dots: {len(trail)}",(10,40),cv2.FONT_HERSHEY_SIMPLEX,1,(255,255,255),2)
    cv2.imshow("Magic Wand Painter",frame)

    # keyboard block
    key = chr(cv2.waitKey(1) & 0xFF)
    if key == "q":
        break
    if key == "c":
        trail = []
    if key in PENS:
        pen = PENS[key]
    if key in SIZES:
        size = SIZES[key]

cap.release()
cv2.destroyAllWindows()
cv2.waitKey(1)
```

```python
def filter_booth():
    COLORS = {"r":(0,0,255),"g":(0,255,0),"b":(255,0,0),"y":(0,255,255)}
    STICKERS = {"1":"hat","2":"glasses","3":"crown","4":"nose","0":"none"}
    color = COLORS["y"]
    sticker = "none"

    cap = cv2.VideoCapture(0)
    print("Colors: r, g, b, y. Stickers: 1, 2, 3, 4, 0. Press q to stop.")

    while True:
        ok,frame = cap.read()
        if not ok:
            break
        frame = cv2.flip(frame,1)

        for x,y,w,h in find_faces(frame):
            if sticker == "hat":
                draw_hat(frame,x,y,w,h,color)
            elif sticker == "glasses":
                draw_glasses(frame,x,y,w,h,color)
            elif sticker == "crown":
                draw_crown(frame,x,y,w,h,color)
            elif sticker == "nose":
                draw_nose(frame,x,y,w,h,color)

        cv2.imshow("Filter Booth",frame)
        key = chr(cv2.waitKey(1) & 0xFF)
        if key == "q":
            break
        if key in COLORS:
            color = COLORS[key]
        if key in STICKERS:
            sticker = STICKERS[key]

    cap.release()
    cv2.destroyAllWindows()
    cv2.waitKey(1)
```

```python
def star_catcher():
    GAME_TIME = 30
    NUMBER_OF_STARS = 4
    player = input("What is your name? ")
    stars = []
    score = 0

    cap = cv2.VideoCapture(0)
    print("Catch gold and diamonds, stay away from the red bombs!")
    start = time.time()

    while True:
        ok,frame = cap.read()
        if not ok:
            break
        frame = cv2.flip(frame,1)
        height,width,channels = frame.shape

        time_left = GAME_TIME - int(time.time() - start)
        if time_left <= 0:
            break

        while len(stars) < NUMBER_OF_STARS:
            stars.append(make_star(width))

        faces = find_faces(frame)
        for x,y,w,h in faces:
            cv2.rectangle(frame,(x,y),(x+w,y+h),(0,255,0),3)

        for star in stars.copy():             # loop over a copy, because we remove stars
            star["y"] = star["y"] + star["speed"]
            cv2.circle(frame,(star["x"],star["y"]),15,star["color"],-1)

            caught = False
            for face in faces:
                if is_inside((star["x"],star["y"]),face):
                    caught = True

            if caught:
                score = score + star["points"]
                stars.remove(star)
            elif star["y"] > height:
                stars.remove(star)

        cv2.putText(frame,f"Score: {score}",(10,40),cv2.FONT_HERSHEY_SIMPLEX,1.2,(255,255,255),3)
        cv2.putText(frame,f"Time: {time_left}",(width-200,40),cv2.FONT_HERSHEY_SIMPLEX,1.2,(255,255,255),3)
        cv2.imshow("Star Catcher",frame)
        key = chr(cv2.waitKey(1) & 0xFF)
        if key == "q":
            break

    cap.release()
    cv2.destroyAllWindows()
    cv2.waitKey(1)

    print(f"Game over! {player}, your score is {score}.")
    if player not in high_scores:
        high_scores[player] = score
    elif score > high_scores[player]:
        high_scores[player] = score
```
