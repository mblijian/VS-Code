```python
import time
import random
sides = random.randint(3,6)
for i in range(sides):
    got.mecanum_move_speed_times(0,30,50,1)
    got.mecanum_turn_speed_times(2,30,int(360/sides),2)
    got.show_light_rgb_effect(0,255,0,3)

print(f"This polygon has {sides} sides.")
got.show_light_rgb_effect(255,0,0,3)
time.sleep(2)
got.turn_off_lights()
got.play_audio_tts("Mission completed", 0, False)
```

```python
import random
A = random.randint(1,10)
B = random.randint(1,10)

if A > B:
    print("A is greater than B")
elif A< B:
    print("A is smaller than B")
else:
    print("A is equal to B")

print(f"A is {A}, B is {B}.")
```
