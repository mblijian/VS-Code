## Session 6

1. while True loop and laser sensor

```python
while True:
    Laser1 = got.read_distance_data(21)
    if Laser1>30:
        got.show_light_rgb_effect(0,255,0,3)
        got.mecanum_move_speed(0,30)
    else:
        got.mecanum_stop()
        got.show_light_rgb_effect(255,0,0,3)
        break
        
```

2. arm control

```python
def pick_up():
    import time
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(1)
    got.mechanical_joint_control(0,0,-70,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(1)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
```

```python
def put_down():
    import time
    got.mechanical_joint_control(-90,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(-90,0,-70,1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(1)
    got.mechanical_joint_control(-90,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(1)
```

3. function

```python
pick_up()
put_down()
```

```python
def move_stop():
    while True:
        Laser1 = got.read_distance_data(21)
        if Laser1>30:
            got.show_light_rgb_effect(0,255,0,3)
            got.mecanum_move_speed(0,30)
        else:
            got.mecanum_stop()
            got.show_light_rgb_effect(255,0,0,3)
            break
```
