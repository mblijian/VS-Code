```python
def pick_up():
    import time
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(0,0,-70,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    
def put_down():
    import time
    got.mechanical_joint_control(-90,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(-90,0,-80,1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(-90,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)

def move_stop():
    while True:
        Laser1 = got.read_distance_data(21)
        if Laser1 > 14:
            got.mecanum_move_speed(0,15)
        else:
            got.mecanum_stop()
            time.sleep(0.5)
            break

def pick_up_100():
    import time
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    move_stop()
    pick_up()

    Laser1 = got.read_distance_data(21)
    while Laser1<12:
        Laser1 = got.read_distance_data(21)
        if Laser1<4:
            got.mecanum_move_speed_times(1,5,6,1)
            time.sleep(1)
            pick_up()
            time.sleep(2)

    print("I have picked up the token.")

def maze_escape():
    # maze escape with while condition loop and the variable move_forward.
    import time
    move_forward = 0
    Laser1 = got.read_distance_data(21)
    while move_forward< 4:
        Laser1 = got.read_distance_data(21)
        if Laser1 >15:
            got.mecanum_move_speed(0,20)
        else:
            got.mecanum_stop()
            move_forward+=1
            print(f"This is move forward: {move_forward}.")
            time.sleep(1)
            got.mecanum_turn_speed_times(3,30,90,2)
            
            Laser1 = got.read_distance_data(21)
            if Laser1<15:
                got.mecanum_turn_speed_times(3,30,180,2)
                time.sleep(0.5)    

def Coley_recognition():
    got.load_models(["face_recognition"])
    face_info = got.get_face_recognition_total_info()

    while True:
        face_info = got.get_face_recognition_total_info()
        if not face_info:
            got.mecanum_move_speed(0,15)
        else:
            got.mecanum_stop()
            break
    got.play_audio_tts("Hello Coley, here's your package, have a nice day!!",0,False)   
```

```python
import random
player_hand = []
dealer_hand = []

# give two cards to player and dealer
for i in range(2):
    player_hand.append(random.randint(1,10))
    dealer_hand.append(random.randint(1,10))

# sort the cards
player_hand.sort()
dealer_hand.sort()

print(f"Your cards : {player_hand}")
print(f"Dealer cards : {dealer_hand}")

while sum(player_hand)<21:
    choice = input("Hit or stand? Enter h or s.").lower()
    if choice =="h":
        new_card = random.randint(1,10)
        player_hand.append(new_card)
        player_hand.sort()

        print(f"Your new card is:{new_card}")
        print(f"Your cards : {player_hand}")
        print(f"Your total points:{sum(player_hand)}")
    elif choice == "s":
        print("You chose to stand.")
        break
    else:
        print("Please enter h or s.")

# compare the player hand and the dealer hand
if sum(player_hand)> 21:
    print("You went over 21, you lose.")
else:
    print(f"Dealer's cards:{dealer_hand}")

    while sum(dealer_hand)< 17:
        new_card = random.randint(1,10)
        dealer_hand.append(new_card)
        dealer_hand.sort()

        print(f"Dealer cards : {dealer_hand}")
        print(f"Dealer total points:{sum(dealer_hand)}")
    #show final result
    player_total = sum(player_hand)
    dealer_total = sum(dealer_hand)

    print(f"Your cards: {player_hand}")
    print(f"Your total:{player_total}")
    print(f"Dealer cards: {dealer_hand}")
    print(f"Dealer total:{dealer_total}")

    # Decide the winner!
    if dealer_total> 21:
        print("The dealer hand went over 21, you win!")
        got.play_audio_tts("The dealer hand went over 21, you win!",0,False)
    elif player_total> dealer_total:
        print("You win!")
        got.play_audio_tts("You win!",1,True)
    elif player_total< dealer_total:
        print("You lose!")
        got.play_audio_tts("You lose!",1,False)
    else:
        print("It is a draw!")
```

```python
import random

# Create three empty player hands
player_hands = [
    [],  # Player 1
    [],  # Player 2
    []   # Player 3
]

dealer_hand = []

# Give two cards to each player
for player_hand in player_hands:
    for i in range(2):
        player_hand.append(random.randint(1, 10))

# Give two cards to the dealer
for i in range(2):
    dealer_hand.append(random.randint(1, 10))

# Sort all the cards
for player_hand in player_hands:
    player_hand.sort()

dealer_hand.sort()

# Show the starting cards
for player_number in range(3):
    print(
        f"Player {player_number + 1}'s cards: "
        f"{player_hands[player_number]}"
    )

print(f"Dealer's cards: {dealer_hand}")


# Each player takes a turn
for player_number in range(3):

    player_hand = player_hands[player_number]

    print()
    print(f"Player {player_number + 1}'s turn!")

    while sum(player_hand) < 21:

        print(f"Your cards: {player_hand}")
        print(f"Your total points: {sum(player_hand)}")

        choice = input("Hit or stand? Enter h or s: ").lower()

        if choice == "h":

            new_card = random.randint(1, 10)
            player_hand.append(new_card)
            player_hand.sort()

            print(f"Your new card is: {new_card}")
            print(f"Your cards: {player_hand}")
            print(f"Your total points: {sum(player_hand)}")

        elif choice == "s":

            print(f"Player {player_number + 1} chose to stand.")
            break

        else:

            print("Please enter h or s.")

    # Check whether the player went over 21
    if sum(player_hand) > 21:
        print(
            f"Player {player_number + 1} went over 21!"
        )

    elif sum(player_hand) == 21:
        print(
            f"Player {player_number + 1} reached 21!"
        )


# Dealer's turn
print()
print("Dealer's turn!")
print(f"Dealer's cards: {dealer_hand}")
print(f"Dealer's total points: {sum(dealer_hand)}")

while sum(dealer_hand) < 17:

    new_card = random.randint(1, 10)
    dealer_hand.append(new_card)
    dealer_hand.sort()

    print(f"Dealer's new card is: {new_card}")
    print(f"Dealer's cards: {dealer_hand}")
    print(f"Dealer's total points: {sum(dealer_hand)}")


# Store the dealer's final total
dealer_total = sum(dealer_hand)

print()
print("Final results")
print(f"Dealer's cards: {dealer_hand}")
print(f"Dealer's total: {dealer_total}")


# Compare each player with the dealer
for player_number in range(3):

    player_hand = player_hands[player_number]
    player_total = sum(player_hand)

    print()
    print(f"Player {player_number + 1}'s cards: {player_hand}")
    print(f"Player {player_number + 1}'s total: {player_total}")

    if player_total > 21:

        result = f"Player {player_number + 1} loses!"

    elif dealer_total > 21:

        result = f"Player {player_number + 1} wins!"

    elif player_total > dealer_total:

        result = f"Player {player_number + 1} wins!"

    elif player_total < dealer_total:

        result = f"Player {player_number + 1} loses!"

    else:

        result = f"Player {player_number + 1} draws with the dealer!"

    print(result)
    got.play_audio_tts(result, 1, False)
```

```python
def take_photo():
    got.open_camera()
    print("The camera is on.")
    
    picture_raw = got.read_camera_data()
    if picture_raw is None:
        print("There is no raw data.")
    else:
        picture_1D  = np.frombuffer(picture_raw,np.uint8)
        picture_bgr = cv2.imdecode(picture_1D,cv2.IMREAD_COLOR)
        ok,picture_jpg = cv2.imencode(".jpg",picture_bgr)
        display(Image(picture_jpg.tobytes()))
```

```python
take_photo()
```

2. guard robot

```python
import time
while True:
    Laser1 = got.read_distance_data(21)
    if Laser1>30:
        got.show_light_rgb_effect(0,255,0,3)
        got.mecanum_turn_speed(2,30)
    else:
        got.mecanum_stop()
        got.show_light_rgb_effect(255,0,0,3)
        take_photo()
        time.sleep(2)
        got.turn_off_lights()
        break
```
