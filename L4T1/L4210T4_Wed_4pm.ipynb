```python
# 1. maze escape function
def maze_escape():
    import time
    move_forward = 0
    while move_forward<4:
        Laser1 = got.read_distance_data(21)
        if Laser1> 10:
            got.mecanum_move_speed(0,20)
        else:
            got.mecanum_stop()
            time.sleep(0.5)
            got.mecanum_turn_speed_times(3,30,90,2)
            move_forward+=1
            print(f"This is move forward:{move_forward}.")
            
            Laser1 = got.read_distance_data(21)
            if Laser1<10:
                got.mecanum_turn_speed_times(2,30,180,2)
                time.sleep(0.5)

# 2. pick up function
def pick_up():
    import time
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30, 1000)
    time.sleep(1)
    got.mechanical_joint_control(0,0,-70, 1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30, 1000)
    time.sleep(1)

# 3. put down function
def put_down():
    import time
    got.mechanical_joint_control(90,30,30, 1000)
    time.sleep(1)
    got.mechanical_joint_control(90,0,-80, 1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(90,30,30, 1000)
    time.sleep(1)
    got.mechanical_joint_control(0,30,30, 1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)

# 4. face recognition function
def face_recognition():
    while True:
        face_info = got.get_face_recognition_total_info()
        if not face_info:
            got.mecanum_move_speed(0,20)
        else:
            got.mecanum_stop()
            break
```

```python
import random

print("Welcome to LUCKY 21!")

player_hand = []
dealer_hand =[]

# give two cards to player and dealer
for i in range(2):
    player_hand.append(random.randint(1,10))
    dealer_hand.append(random.randint(1,10))

# sort the cards from smallest to largest
player_hand.sort()
dealer_hand.sort()

print(f"Your cards : {player_hand}")
print(f"Your total:{sum(player_hand)}")

while sum(player_hand)< 21:
    choice = input("Hit or stand?: Enter h or s").lower()
    if choice == "h":
        new_card = random.randint(1,10)
        player_hand.append(new_card)
        player_hand.sort()

        print(f"You receive:{new_card}")
        print(f"Your cards : {player_hand}")
        print(f"Your total:{sum(player_hand)}")
    
    elif choice =="s":
        print("You chose to stand.")
        break
    else:
        print("Please enter h or s.")

# compare the dealer hand and player hand
if sum(player_hand)> 21:
    print("You went over 21, you loose.")

else:
    print(f"Dealer's cards:{dealer_hand} ")

    while sum(dealer_hand)<17:
        new_card = random.randint(1,10)
        dealer_hand.append(new_card)
        dealer_hand.sort()

        print(f"Your cards : {dealer_hand}")
        print(f"Your total:{sum(dealer_hand)}")
    # show final result
    player_total = sum(player_hand)
    dealer_total = sum(dealer_hand)

    print("FINAL RESULT!")
    print(f"Your cards: {player_hand}")
    print(f"Your total: {sum(player_hand)}")
    print(f"Dealer cards: {dealer_hand}")
    print(f"Dealer total: {sum(dealer_hand)}")

    # Decide the winner!
    if dealer_total>21:
        print("The dealer went over 21. You win!")
        got.play_audio_tts("The dealer went over 21. You win!",1,False)
    elif player_total>dealer_total:
        print("The player wins!")
        got.play_audio_tts("The player wins!",1,False)
    elif player_total<dealer_total:
        print("The dealer wins!")
        got.play_audio_tts("The dealer wins!",1,False)
    else:
        print("It is a draw!")
        got.play_audio_tts("It is a draw!",1,False)
```

2. diagonal move

```python
def face_diagonal_centralize():
    import time
    got.load_models(["face_recognition"])
    time.sleep(1)

    while True:
        face_info = got.get_face_recognition_total_info()
        if face_info:
            face_x = face_info[0][1]
            face_width = face_info[0][4]

            # Face is close enough and centered
            if face_width > 70 and 300 < face_x < 380:
                got.mecanum_stop()
                print("Face centered!")
                break

            # Face is on the left → move diagonally left + forward
            elif face_x < 300:
                got.mecanum_move_xyz(-20, 10, 0)

            # Face is on the right → move diagonally right + forward
            elif face_x > 380:
                got.mecanum_move_xyz(20, 10, 0)

            # Face is centered, but still too far away
            else:
                got.mecanum_move_xyz(0, 10, 0)

        else:
            # No face → move forward to search
            got.mecanum_move_xyz(0, 10, 0)

        time.sleep(0.1)
```

4. color cube diagnoal approach

```python
# get ready block
import time
got.load_models(["color_recognition"])
got.open_camera()
print("The camera is on.")

while True:    
    # take photo block
    picture_raw = got.read_camera_data()
    if picture_raw is None:
        continue
    frame = cv2.imdecode(np.frombuffer(picture_raw,np.uint8),cv2.IMREAD_COLOR)
    
    # bounding box block
    color_info = got.get_color_total_info()
    if color_info:
        color,shape,cx,cy,height,width,area = got.get_color_total_info()
        cx,cy,height,width = map(int,(cx,cy,height,width))
        cv2.rectangle(frame,(cx-width//2,cy-height//2),(cx+width//2,cy+height//2),(0,255,0),3)
        cv2.putText(frame,f"{color} {shape}",(cx-30,cy-80),cv2.FONT_HERSHEY_SIMPLEX,0.6,(0,0,255),2)
        cv2.circle(frame,(cx,cy),3,(0,255,0),3)
        
    # diagnoal move block
        if width>90 and 280<cx<380:
            got.mecanum_stop()
            break
        elif cx > 380: # face too much right
            got.mecanum_move_xyz(10,5,0)
        elif cx < 280: # face too much left
            got.mecanum_move_xyz(-10,5,0)
        else: # face in the middle
            got.mecanum_move_xyz(0,5,0)
    else:
        got.mecanum_move_xyz(0, 10, 0)
        
    # show the video block
    ok,picture_jpg = cv2.imencode(".jpg",frame)
    display(Image(picture_jpg.tobytes()))
    clear_output(True)
```

```python
# 1. cube recognition
def color_cube_recognition():
    # get ready block
    import time
    got.load_models(["color_recognition"])
    got.open_camera()
    print("The camera is on.")

    while True:    
        # take photo block
        picture_raw = got.read_camera_data()
        if picture_raw is None:
            continue
        frame = cv2.imdecode(np.frombuffer(picture_raw,np.uint8),cv2.IMREAD_COLOR)
        
        # bounding box block
        color_info = got.get_color_total_info()
        if color_info:
            color,shape,cx,cy,height,width,area = got.get_color_total_info()
            cx,cy,height,width = map(int,(cx,cy,height,width))
            cv2.rectangle(frame,(cx-width//2,cy-height//2),(cx+width//2,cy+height//2),(0,255,0),3)
            cv2.putText(frame,f"{color} {shape}",(cx-30,cy-80),cv2.FONT_HERSHEY_SIMPLEX,0.6,(0,0,255),2)
            cv2.circle(frame,(cx,cy),3,(0,255,0),3)
            
        # diagnoal move block
            if width>110 and 280<cx<380:
                got.mecanum_stop()
                break
            elif cx > 380: # face too much right
                got.mecanum_move_xyz(10,5,0)
            elif cx < 280: # face too much left
                got.mecanum_move_xyz(-10,5,0)
            else: # face in the middle
                got.mecanum_move_xyz(0,5,0)
        else:
            got.mecanum_move_xyz(0, 10, 0)
            
        # show the video block
        ok,picture_jpg = cv2.imencode(".jpg",frame)
        display(Image(picture_jpg.tobytes()))
        clear_output(True)
```
