```python
# 1. pick up
def pick_up():
    import time
    got.mechanical_clamp_release()
    time.sleep(2)
    got.load_models(["apriltag_qrcode"])
    AP_info = got.get_apriltag_total_info()
    AP_x = AP_info[0][1]

    for i in range(10):
        AP_info = got.get_apriltag_total_info()
        AP_x = AP_info[0][1]
        joint1 = int((320-AP_x)*0.1)
        got.mechanical_joint_control(joint1,30,30,1000)
        time.sleep(0.2)
    got.mechanical_joint_control(joint1,0,-70,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)
    got.mechanical_joint_control(-90,30,30,1000)
    time.sleep(1)

# 2. put down    
def put_down():
    import time
    got.mechanical_joint_control(-90,30,30,1000)
    time.sleep(1)
    got.mechanical_joint_control(0,0,-80,1000)
    time.sleep(1)
    got.mechanical_clamp_release()
    time.sleep(2)
    got.mechanical_joint_control(0,30,30,1000)
    time.sleep(1)
    got.mechanical_clamp_close()
    time.sleep(2)

# 3. line follow
def line_follow():
    import time
    got.load_models(["line_recognition"])
    got.set_track_recognition_line(0)

    while True:
        line_info =got.get_single_track_total_info()
        offset = line_info[0]
        line_type = line_info[1]

        if line_type> 0:
            got.mecanum_move_xyz(0,20,int(0.2*offset))
            time.sleep(0.1)
        else:
            got.mecanum_stop()
            break

    print("There's no line to follow, let's stop!")

# 4. face recognition
def face_recognition():
    got.load_models(["face_recognition"])
    while True:
        face_info = got.get_face_recognition_total_info()
        if face_info:
            face_x = face_info[0][1]
            face_width = face_info[0][4]

            # if face is colse enough and centered
            if face_width >70 and 280<face_x<360:
                got.mecanum_stop()
                break

            # face is on the left-move left
            elif face_x<280:
                got.mecanum_move_xyz(-15,5,0)

            # face is on the right-move right
            elif face_x>360:
                got.mecanum_move_xyz(15,5,0)

            # face is in the middle, move forward
            else:
                got.mecanum_move_xyz(0,5,0)
        else:
            got.mecanum_move_xyz(0,20,0)

# 5. AP recognition
def AP_recognition():
    got.load_models(["apriltag_qrcode"])
    while True:
        AP_info = got.get_apriltag_total_info()
        if AP_info:
            AP_x = AP_info[0][1]
            AP_distance = AP_info[0][6]
            # if AP is colse enough and centered
            if AP_distance <0.2 and 280<AP_x<360:
                got.mecanum_stop()
                break
            # AP is on the left-move left
            elif AP_x<280:
                got.mecanum_move_xyz(-10,5,0)
            # AP is on the right-move right
            elif AP_x>360:
                got.mecanum_move_xyz(10,5,0)
            # AP is in the middle, move forward
            else:
                got.mecanum_move_xyz(0,5,0)
        else:
            got.mecanum_move_xyz(0,20,0)

# 6. line detect
def line_detect():
    got.load_models(["line_recognition"])
    got.set_track_recognition_line(0)

    while True:
        line_info =got.get_single_track_total_info()
        line_type = line_info[1]
        if line_type !=0:
            got.mecanum_stop()
            break
        got.mecanum_move_xyz(0,10,0)
```

```python
import time
AP_recognition()
time.sleep(1)
pick_up()
time.sleep(2)
line_follow()
time.sleep(2)
face_recognition()
time.sleep(1)
put_down()
time.sleep(1)
got.mecanum_turn_speed_times(2,30,180,2)
got.mechanical_joint_control(-90,30,30,1000)
time.sleep(1)
line_detect()
line_follow()
```
